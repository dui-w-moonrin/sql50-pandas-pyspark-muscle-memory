# 0. Problem
## 610. Triangle Judgement — Easy
Return `Yes` when x, y, z can form a triangle; otherwise `No`.
Official: https://leetcode.com/problems/triangle-judgement/

# 1. Setup

```python
import pandas as pd
import numpy as np
triangle_rows=[(13,15,30),(10,20,15),(5,5,10),(7,7,7)]
triangle_pd=pd.DataFrame(triangle_rows,columns=["x","y","z"])
triangle_pd
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
spark=SparkSession.builder.getOrCreate()
triangle_spark=spark.createDataFrame(triangle_rows,["x","y","z"])
triangle_spark.createOrReplaceTempView("Triangle")
```

# 2. SQL Solution

```python
sql_result=spark.sql("""SELECT x,y,z,CASE WHEN x+y>z AND x+z>y AND y+z>x THEN 'Yes' ELSE 'No' END AS triangle FROM Triangle""")
sql_result.show(truncate=False)
```

# 3. pandas Solution

```python
valid=(triangle_pd["x"]+triangle_pd["y"]>triangle_pd["z"])&(triangle_pd["x"]+triangle_pd["z"]>triangle_pd["y"])&(triangle_pd["y"]+triangle_pd["z"]>triangle_pd["x"])
result_pd=triangle_pd.assign(triangle=np.where(valid,"Yes","No"))
result_pd
```

# 4. PySpark Solution

```python
valid=(F.col("x")+F.col("y")>F.col("z"))&(F.col("x")+F.col("z")>F.col("y"))&(F.col("y")+F.col("z")>F.col("x"))
result_spark=triangle_spark.withColumn("triangle",F.when(valid,F.lit("Yes")).otherwise(F.lit("No")))
result_spark.show(truncate=False)
```

# 5. Pattern Mapping
| Concept | SQL | pandas | PySpark |
|---|---|---|---|
| conditional output | `CASE WHEN` | `np.where()` | `F.when().otherwise()` |
| AND | `AND` | `&` | `&` |
