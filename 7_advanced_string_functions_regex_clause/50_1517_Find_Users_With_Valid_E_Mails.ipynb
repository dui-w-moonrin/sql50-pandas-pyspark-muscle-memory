# 0. Problem
## 1517. Find Users With Valid E-Mails — Easy
A valid email starts with a letter, may continue with letters/digits/underscore/dot/dash, and ends exactly with `@leetcode.com`.

Official: https://leetcode.com/problems/find-users-with-valid-e-mails/

# 1. Setup

```python
import pandas as pd
users_rows=[(1,"Winston","winston@leetcode.com"),(2,"Jonathan","jonathanisgreat"),(3,"Annabelle","bella-@leetcode.com"),(4,"Sally","sally.come@leetcode.com"),(5,"Marwan","quarz#2020@leetcode.com"),(6,"David","david69@gmail.com"),(7,"Shapiro",".shapo@leetcode.com"),(8,"Amy","A_b.c-d@leetcode.com")]
users_pd=pd.DataFrame(users_rows,columns=["user_id","name","mail"])
users_pd
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
spark=SparkSession.builder.getOrCreate()
users_spark=spark.createDataFrame(users_rows,["user_id","name","mail"])
users_spark.createOrReplaceTempView("Users")
```

# 2. SQL Solution

```python
sql_result=spark.sql("""
SELECT user_id,name,mail
FROM Users
WHERE mail RLIKE '^[A-Za-z][A-Za-z0-9_.-]*@leetcode[.]com$'
ORDER BY user_id
""")
sql_result.show(truncate=False)
```

# 3. pandas Solution

```python
pattern=r"^[A-Za-z][A-Za-z0-9_.-]*@leetcode[.]com$"
result_pd=users_pd.loc[users_pd["mail"].str.fullmatch(pattern,na=False)].sort_values("user_id").reset_index(drop=True)
result_pd
```

# 4. PySpark Solution

```python
pattern=r"^[A-Za-z][A-Za-z0-9_.-]*@leetcode[.]com$"
result_spark=users_spark.filter(F.col("mail").rlike(pattern)).orderBy("user_id")
result_spark.show(truncate=False)
```

# 5. Pattern Mapping
| Concept | SQL | pandas | PySpark |
|---|---|---|---|
| full email regex | `RLIKE ^...$` | `.str.fullmatch()` | `.rlike()` |
| allowed prefix | character class | same regex | same regex |
