# 0. Problem
## 1667. Fix Names in a Table — Easy
Capitalize the first character of each name and lowercase the rest. Sort by `user_id`.
Official: https://leetcode.com/problems/fix-names-in-a-table/

# 1. Setup

```python
import pandas as pd
users_rows=[(1,"aLice"),(2,"bOB"),(3,"cHArLie")]
users_pd=pd.DataFrame(users_rows,columns=["user_id","name"])
users_pd
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
spark=SparkSession.builder.getOrCreate()
users_spark=spark.createDataFrame(users_rows,["user_id","name"])
users_spark.createOrReplaceTempView("Users")
```

# 2. SQL Solution

```python
sql_result=spark.sql("""SELECT user_id,CONCAT(UPPER(SUBSTRING(name,1,1)),LOWER(SUBSTRING(name,2,LENGTH(name)))) AS name FROM Users ORDER BY user_id""")
sql_result.show(truncate=False)
```

# 3. pandas Solution

```python
result_pd=users_pd.assign(name=users_pd["name"].str.capitalize()).sort_values("user_id").reset_index(drop=True)
result_pd
```

# 4. PySpark Solution

```python
result_spark=(users_spark.withColumn("name",F.concat(F.upper(F.substring("name",1,1)),F.lower(F.substring("name",2,1000)))).orderBy("user_id"))
result_spark.show(truncate=False)
```

# 5. Pattern Mapping
| Concept | SQL | pandas | PySpark |
|---|---|---|---|
| capitalize | `UPPER/LOWER/SUBSTRING` | `.str.capitalize()` | `upper/lower/substring/concat` |
