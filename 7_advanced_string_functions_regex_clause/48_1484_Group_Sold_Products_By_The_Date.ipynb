# 0. Problem
## 1484. Group Sold Products By The Date — Easy
For each sell date, return the number of distinct products and a comma-separated alphabetically sorted product list.
Official: https://leetcode.com/problems/group-sold-products-by-the-date/

# 1. Setup

```python
import pandas as pd
activities_rows=[("2020-05-30","Headphone"),("2020-06-01","Pencil"),("2020-06-02","Mask"),("2020-05-30","Basketball"),("2020-06-01","Bible"),("2020-06-02","Mask"),("2020-05-30","T-Shirt")]
activities_pd=pd.DataFrame(activities_rows,columns=["sell_date","product"]); activities_pd
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
spark=SparkSession.builder.getOrCreate(); activities_spark=spark.createDataFrame(activities_rows,["sell_date","product"]); activities_spark.createOrReplaceTempView("Activities")
```

# 2. SQL Solution

```python
sql_result=spark.sql("""SELECT sell_date,COUNT(DISTINCT product) AS num_sold,CONCAT_WS(',',SORT_ARRAY(COLLECT_SET(product))) AS products FROM Activities GROUP BY sell_date ORDER BY sell_date"""); sql_result.show(truncate=False)
```

# 3. pandas Solution

```python
result_pd=(activities_pd.groupby("sell_date",as_index=False).agg(num_sold=("product","nunique"),products=("product",lambda s:",".join(sorted(set(s))))).sort_values("sell_date").reset_index(drop=True)); result_pd
```

# 4. PySpark Solution

```python
result_spark=(activities_spark.groupBy("sell_date").agg(F.countDistinct("product").alias("num_sold"),F.concat_ws(",",F.sort_array(F.collect_set("product"))).alias("products")).orderBy("sell_date")); result_spark.show(truncate=False)
```

# 5. Pattern Mapping
| Concept | SQL | pandas | PySpark |
|---|---|---|---|
| distinct count | `COUNT(DISTINCT)` | `.nunique()` | `countDistinct()` |
| sorted string aggregation | `SORT_ARRAY(COLLECT_SET)` | `join(sorted(set(...)))` | `concat_ws(sort_array(collect_set))` |
