# 1729. Find Followers Count

[LeetCode problem](https://leetcode.com/problems/find-followers-count/)


# 0. Problem

Count how many followers each `user_id` has. Return results ordered by `user_id`.


# 1. Setup


```python
import pandas as pd
follower_rows=[(0,1),(1,0),(2,0),(2,1),(3,0),(3,1),(3,2)]
followers_pd=pd.DataFrame(follower_rows,columns=['user_id','follower_id'])
```

```python
# In Colab if needed: !pip -q install pyspark
from pyspark.sql import SparkSession, functions as F, types as T
spark=SparkSession.builder.getOrCreate()
schema=T.StructType([T.StructField('user_id',T.IntegerType(),False),T.StructField('follower_id',T.IntegerType(),False)])
followers_spark=spark.createDataFrame(follower_rows,schema)
followers_spark.createOrReplaceTempView('Followers')
```

# 2. SQL Solution


```python
sql_result=spark.sql("""SELECT user_id, COUNT(follower_id) AS followers_count FROM Followers GROUP BY user_id ORDER BY user_id""")
sql_result.show(truncate=False)
```

# 3. pandas Solution


```python
pandas_result=followers_pd.groupby('user_id',as_index=False).agg(followers_count=('follower_id','count')).sort_values('user_id').reset_index(drop=True)
pandas_result
```

# 4. PySpark Solution


```python
spark_result=followers_spark.groupBy('user_id').agg(F.count('follower_id').alias('followers_count')).orderBy('user_id')
spark_result.show(truncate=False)
```

# 5. Pattern Mapping

| Concept | SQL | pandas | PySpark |
|---|---|---|---|
| group | `GROUP BY` | `.groupby()` | `.groupBy()` |
| count | `COUNT()` | `.agg(...,'count')` | `F.count()` |
| sort | `ORDER BY` | `.sort_values()` | `.orderBy()` |
