# 0. Problem
## 619. Biggest Single Number — Easy
Find the largest number that appears exactly once. Return NULL/None if no such number exists.
Official: https://leetcode.com/problems/biggest-single-number/

# 1. Setup

```python
import pandas as pd
numbers_rows=[(8,),(8,),(3,),(3,),(1,),(4,),(5,),(6,)]; numbers_pd=pd.DataFrame(numbers_rows,columns=["num"]); numbers_pd
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
spark=SparkSession.builder.getOrCreate(); numbers_spark=spark.createDataFrame(numbers_rows,["num"]); numbers_spark.createOrReplaceTempView("MyNumbers")
```

# 2. SQL Solution

```python
sql_result=spark.sql("""SELECT MAX(num) AS num FROM (SELECT num FROM MyNumbers GROUP BY num HAVING COUNT(*)=1) x"""); sql_result.show(truncate=False)
```

# 3. pandas Solution

```python
counts=numbers_pd["num"].value_counts(); singles=counts[counts.eq(1)].index.tolist(); result_pd=pd.DataFrame({"num":[max(singles) if singles else None]}); result_pd
```

# 4. PySpark Solution

```python
result_spark=(numbers_spark.groupBy("num").count().filter(F.col("count")==1).agg(F.max("num").alias("num"))); result_spark.show(truncate=False)
```

# 5. Pattern Mapping
| Concept | SQL | pandas | PySpark |
|---|---|---|---|
| appears once | `HAVING COUNT(*)=1` | `.value_counts()` | `.groupBy().count().filter()` |
| maximum | `MAX()` | `max()` | `F.max()` |
