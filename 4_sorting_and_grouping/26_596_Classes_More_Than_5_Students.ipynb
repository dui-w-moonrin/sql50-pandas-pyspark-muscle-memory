# 0. Problem
## 596. Classes More Than 5 Students — Easy
Return classes with at least 5 students.

Official: https://leetcode.com/problems/classes-more-than-5-students/

# 1. Setup

```python
import pandas as pd
courses_rows=[("A","Math"),("B","English"),("C","Math"),("D","Biology"),("E","Math"),("F","Computer"),("G","Math"),("H","Math"),("I","English"),("J","Math")]
courses_pd=pd.DataFrame(courses_rows,columns=["student","class"])
courses_pd
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
spark=SparkSession.builder.getOrCreate()
courses_spark=spark.createDataFrame(courses_rows,["student","class"])
courses_spark.createOrReplaceTempView("Courses")
```

# 2. SQL Solution

```python
sql_result=spark.sql("""
SELECT class
FROM Courses
GROUP BY class
HAVING COUNT(student)>=5
ORDER BY class
""")
sql_result.show(truncate=False)
```

# 3. pandas Solution

```python
counts_pd=courses_pd.groupby("class").size().rename("student_count").reset_index()
result_pd=counts_pd.loc[counts_pd["student_count"]>=5,["class"]].sort_values("class").reset_index(drop=True)
result_pd
```

# 4. PySpark Solution

```python
result_spark=(courses_spark.groupBy("class").agg(F.count("student").alias("student_count")).filter(F.col("student_count")>=5).select("class").orderBy("class"))
result_spark.show(truncate=False)
```

# 5. Pattern Mapping
| Concept | SQL | pandas | PySpark |
|---|---|---|---|
| qualify groups | `HAVING` | aggregate then boolean filter | aggregate then `.filter()` |
