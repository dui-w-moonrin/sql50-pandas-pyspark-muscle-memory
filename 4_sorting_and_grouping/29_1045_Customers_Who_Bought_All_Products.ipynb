# 0. Problem
## 1045. Customers Who Bought All Products — Medium
Return customers who have purchased every product in the Product table.
Official: https://leetcode.com/problems/customers-who-bought-all-products/

# 1. Setup

```python
import pandas as pd
customer_rows=[(1,5),(2,6),(3,5),(3,6),(1,6)]; product_rows=[(5,),(6,)]; customer_pd=pd.DataFrame(customer_rows,columns=["customer_id","product_key"]); product_pd=pd.DataFrame(product_rows,columns=["product_key"])
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
spark=SparkSession.builder.getOrCreate(); customer_spark=spark.createDataFrame(customer_rows,["customer_id","product_key"]); product_spark=spark.createDataFrame(product_rows,["product_key"]); customer_spark.createOrReplaceTempView("Customer"); product_spark.createOrReplaceTempView("Product")
```

# 2. SQL Solution

```python
sql_result=spark.sql("""SELECT customer_id FROM Customer GROUP BY customer_id HAVING COUNT(DISTINCT product_key)=(SELECT COUNT(*) FROM Product) ORDER BY customer_id"""); sql_result.show(truncate=False)
```

# 3. pandas Solution

```python
total_products=product_pd["product_key"].nunique(); counts=customer_pd.groupby("customer_id",as_index=False).agg(products_bought=("product_key","nunique")); result_pd=counts.loc[counts["products_bought"].eq(total_products),["customer_id"]].sort_values("customer_id").reset_index(drop=True); result_pd
```

# 4. PySpark Solution

```python
total_products=product_spark.select("product_key").distinct().count(); result_spark=(customer_spark.groupBy("customer_id").agg(F.countDistinct("product_key").alias("products_bought")).filter(F.col("products_bought")==F.lit(total_products)).select("customer_id").orderBy("customer_id")); result_spark.show(truncate=False)
```

# 5. Pattern Mapping
| Concept | SQL | pandas | PySpark |
|---|---|---|---|
| relational division / ALL | distinct bought count = product count | group nunique = total | countDistinct = total |
