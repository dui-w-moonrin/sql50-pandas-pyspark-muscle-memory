# 1378. Replace Employee ID With The Unique Identifier

[LeetCode problem](https://leetcode.com/problems/replace-employee-id-with-the-unique-identifier/)


# 0. Problem

Return every employee name with its `unique_id`. Employees without a match must still appear with null.


# 1. Setup


```python
import pandas as pd
employee_rows=[(1,'Alice'),(7,'Bob'),(11,'Meir'),(90,'Winston'),(3,'Jonathan')]
uni_rows=[(3,1),(11,2),(90,3)]
employees_pd=pd.DataFrame(employee_rows,columns=['id','name'])
employee_uni_pd=pd.DataFrame(uni_rows,columns=['id','unique_id'])
```

```python
# In Colab if needed: !pip -q install pyspark
from pyspark.sql import SparkSession, types as T
spark=SparkSession.builder.getOrCreate()
employee_schema=T.StructType([T.StructField('id',T.IntegerType(),False),T.StructField('name',T.StringType(),False)])
uni_schema=T.StructType([T.StructField('id',T.IntegerType(),False),T.StructField('unique_id',T.IntegerType(),False)])
employees_spark=spark.createDataFrame(employee_rows,employee_schema)
employee_uni_spark=spark.createDataFrame(uni_rows,uni_schema)
employees_spark.createOrReplaceTempView('Employees')
employee_uni_spark.createOrReplaceTempView('EmployeeUNI')
```

# 2. SQL Solution


```python
sql_result=spark.sql("""SELECT u.unique_id,e.name FROM Employees e LEFT JOIN EmployeeUNI u ON e.id=u.id""")
sql_result.show(truncate=False)
```

# 3. pandas Solution


```python
pandas_result=employees_pd.merge(employee_uni_pd,on='id',how='left')[['unique_id','name']]
pandas_result
```

# 4. PySpark Solution


```python
spark_result=employees_spark.alias('e').join(employee_uni_spark.alias('u'),on='id',how='left').select('u.unique_id','e.name')
spark_result.show(truncate=False)
```

# 5. Pattern Mapping

| Concept | SQL | pandas | PySpark |
|---|---|---|---|
| left join | `LEFT JOIN` | `.merge(...,how='left')` | `.join(...,how='left')` |
| projection | `SELECT` | `[[...]]` | `.select()` |
