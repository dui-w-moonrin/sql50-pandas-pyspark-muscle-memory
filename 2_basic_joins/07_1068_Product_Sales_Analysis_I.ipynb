# 0. Problem
## 1068. Product Sales Analysis I — Easy
Join sales to product master and return `product_name`, `year`, and `price`.
Official: https://leetcode.com/problems/product-sales-analysis-i/

# 1. Setup

```python
import pandas as pd
sales_rows=[(1,100,2008,10,5000),(2,100,2009,12,5000),(7,200,2011,15,9000)]
product_rows=[(100,"Nokia"),(200,"Apple"),(300,"Samsung")]
sales_pd=pd.DataFrame(sales_rows,columns=["sale_id","product_id","year","quantity","price"])
product_pd=pd.DataFrame(product_rows,columns=["product_id","product_name"])
```

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.getOrCreate()
sales_spark=spark.createDataFrame(sales_rows,["sale_id","product_id","year","quantity","price"])
product_spark=spark.createDataFrame(product_rows,["product_id","product_name"])
sales_spark.createOrReplaceTempView("Sales")
product_spark.createOrReplaceTempView("Product")
```

# 2. SQL Solution

```python
sql_result=spark.sql("""SELECT p.product_name,s.year,s.price FROM Sales s JOIN Product p ON s.product_id=p.product_id""")
sql_result.show(truncate=False)
```

# 3. pandas Solution

```python
result_pd=sales_pd.merge(product_pd,on="product_id",how="inner")[["product_name","year","price"]]
result_pd
```

# 4. PySpark Solution

```python
result_spark=sales_spark.join(product_spark,on="product_id",how="inner").select("product_name","year","price")
result_spark.show(truncate=False)
```

# 5. Pattern Mapping
| Concept | SQL | pandas | PySpark |
|---|---|---|---|
| inner join | `JOIN ... ON` | `.merge(..., how="inner")` | `.join(..., how="inner")` |
