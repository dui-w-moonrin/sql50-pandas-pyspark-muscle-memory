# 0. Problem
## 577. Employee Bonus — Easy
Return employee `name` and `bonus` where bonus < 1000, including employees with no bonus row.
Official: https://leetcode.com/problems/employee-bonus/

# 1. Setup

```python
import pandas as pd
employee_rows=[(3,"Brad",None,4000),(1,"John",3,1000),(2,"Dan",3,2000),(4,"Thomas",3,4000)]
bonus_rows=[(2,500),(4,2000)]
employee_pd=pd.DataFrame(employee_rows,columns=["empId","name","supervisor","salary"])
bonus_pd=pd.DataFrame(bonus_rows,columns=["empId","bonus"])
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
spark=SparkSession.builder.getOrCreate()
employee_spark=spark.createDataFrame(employee_rows,"empId int, name string, supervisor int, salary int")
bonus_spark=spark.createDataFrame(bonus_rows,["empId","bonus"])
employee_spark.createOrReplaceTempView("Employee")
bonus_spark.createOrReplaceTempView("Bonus")
```

# 2. SQL Solution

```python
sql_result=spark.sql("""SELECT e.name,b.bonus FROM Employee e LEFT JOIN Bonus b ON e.empId=b.empId WHERE b.bonus<1000 OR b.bonus IS NULL""")
sql_result.show(truncate=False)
```

# 3. pandas Solution

```python
joined_pd=employee_pd.merge(bonus_pd,on="empId",how="left")
result_pd=joined_pd.loc[joined_pd["bonus"].lt(1000)|joined_pd["bonus"].isna(),["name","bonus"]]
result_pd
```

# 4. PySpark Solution

```python
result_spark=(employee_spark.join(bonus_spark,on="empId",how="left").filter((F.col("bonus")<1000)|F.col("bonus").isNull()).select("name","bonus"))
result_spark.show(truncate=False)
```

# 5. Pattern Mapping
| Concept | SQL | pandas | PySpark |
|---|---|---|---|
| preserve unmatched | `LEFT JOIN` | `how="left"` | `how="left"` |
| null check | `IS NULL` | `.isna()` | `.isNull()` |
