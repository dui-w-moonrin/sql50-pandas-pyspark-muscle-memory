# 0. Problem
## 595. Big Countries — Easy
Return `name`, `population`, and `area` when area >= 3,000,000 or population >= 25,000,000.
Official: https://leetcode.com/problems/big-countries/

# 1. Setup

```python
import pandas as pd
world_rows=[("Afghanistan","Asia",652230,25500100,20343000000),("Albania","Europe",28748,2831741,12960000000),("Algeria","Africa",2381741,37100000,188681000000),("Andorra","Europe",468,78115,3712000000),("Angola","Africa",1246700,20609294,100990000000)]
world_pd=pd.DataFrame(world_rows,columns=["name","continent","area","population","gdp"])
world_pd
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
spark=SparkSession.builder.getOrCreate()
world_spark=spark.createDataFrame(world_rows,["name","continent","area","population","gdp"])
world_spark.createOrReplaceTempView("World")
```

# 2. SQL Solution

```python
sql_result=spark.sql("""SELECT name,population,area FROM World WHERE area>=3000000 OR population>=25000000""")
sql_result.show(truncate=False)
```

# 3. pandas Solution

```python
result_pd=world_pd.loc[(world_pd["area"]>=3_000_000)|(world_pd["population"]>=25_000_000),["name","population","area"]]
result_pd
```

# 4. PySpark Solution

```python
result_spark=world_spark.filter((F.col("area")>=3_000_000)|(F.col("population")>=25_000_000)).select("name","population","area")
result_spark.show(truncate=False)
```

# 5. Pattern Mapping
| Concept | SQL | pandas | PySpark |
|---|---|---|---|
| OR filter | `OR` | `|` | `|` |
| projection | `SELECT` | `.loc[..., cols]` | `.select()` |
