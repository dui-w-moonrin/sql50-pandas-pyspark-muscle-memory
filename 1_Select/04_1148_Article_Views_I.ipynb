# 0. Problem
## 1148. Article Views I — Easy
Find authors who viewed their own articles. Return unique IDs as `id`, sorted ascending.
Official: https://leetcode.com/problems/article-views-i/

# 1. Setup

```python
import pandas as pd
views_rows=[(1,3,5,"2019-08-01"),(1,3,6,"2019-08-02"),(2,7,7,"2019-08-01"),(2,7,6,"2019-08-02"),(4,7,1,"2019-07-22"),(3,4,4,"2019-07-21"),(3,4,4,"2019-07-21")]
views_pd=pd.DataFrame(views_rows,columns=["article_id","author_id","viewer_id","view_date"])
views_pd
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
spark=SparkSession.builder.getOrCreate()
views_spark=spark.createDataFrame(views_rows,["article_id","author_id","viewer_id","view_date"])
views_spark.createOrReplaceTempView("Views")
```

# 2. SQL Solution

```python
sql_result=spark.sql("""SELECT DISTINCT author_id AS id FROM Views WHERE author_id=viewer_id ORDER BY id""")
sql_result.show(truncate=False)
```

# 3. pandas Solution

```python
result_pd=(views_pd.loc[views_pd["author_id"]==views_pd["viewer_id"],["author_id"]].drop_duplicates().rename(columns={"author_id":"id"}).sort_values("id").reset_index(drop=True))
result_pd
```

# 4. PySpark Solution

```python
result_spark=(views_spark.filter(F.col("author_id")==F.col("viewer_id")).select(F.col("author_id").alias("id")).distinct().orderBy("id"))
result_spark.show(truncate=False)
```

# 5. Pattern Mapping
| Concept | SQL | pandas | PySpark |
|---|---|---|---|
| compare columns | `a=b` | `df[a]==df[b]` | `F.col(a)==F.col(b)` |
| unique | `DISTINCT` | `.drop_duplicates()` | `.distinct()` |
| sort | `ORDER BY` | `.sort_values()` | `.orderBy()` |
