# 1757. Recyclable and Low Fat Products

[LeetCode problem](https://leetcode.com/problems/recyclable-and-low-fat-products/)


# 0. Problem

Return the `product_id` values whose rows have both `low_fats = Y` and `recyclable = Y`.

Expected result for the mock data: product IDs `1` and `3`.


# 1. Setup

Create deterministic mock data for pandas, PySpark, and Spark SQL.


```python
import pandas as pd

product_rows = [
    (0, 'Y', 'N'),
    (1, 'Y', 'Y'),
    (2, 'N', 'Y'),
    (3, 'Y', 'Y'),
    (4, 'N', 'N'),
]

product_pd = pd.DataFrame(product_rows, columns=['product_id', 'low_fats', 'recyclable'])
```

```python
# PySpark / Spark SQL setup
# In Google Colab, install once if needed: !pip -q install pyspark
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql import types as T
from pyspark.sql.window import Window

spark = SparkSession.builder.getOrCreate()

product_schema = T.StructType([
    T.StructField('product_id', T.IntegerType(), False),
    T.StructField('low_fats', T.StringType(), False),
    T.StructField('recyclable', T.StringType(), False),
])
product_spark = spark.createDataFrame(product_rows, product_schema)
product_spark.createOrReplaceTempView('Products')

product_pd
```

# 2. SQL Solution


```python
sql_result = spark.sql("""
SELECT product_id
FROM Products
WHERE low_fats = 'Y'
  AND recyclable = 'Y'
""")
sql_result.show(truncate=False)
```

# 3. pandas Solution


```python
pandas_result = product_pd.loc[
    (product_pd['low_fats'] == 'Y') & (product_pd['recyclable'] == 'Y'),
    ['product_id'],
]
pandas_result
```

# 4. PySpark Solution


```python
spark_result = (
    product_spark
    .filter((F.col('low_fats') == 'Y') & (F.col('recyclable') == 'Y'))
    .select('product_id')
)
spark_result.show(truncate=False)
```

# 5. Pattern Mapping

| Concept | SQL | pandas | PySpark |
|---|---|---|---|
| equality | `=` | `==` | `==` |
| AND | `AND` | `&` | `&` |
| filter | `WHERE` | `.loc[...]` | `.filter()` |
| projection | `SELECT product_id` | `[['product_id']]` / `.loc[..., ['product_id']]` | `.select('product_id')` |
